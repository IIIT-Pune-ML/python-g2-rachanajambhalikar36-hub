```
#use of a iterator
#iterator- returns numbers that starts with 25 and each and every time it is passed it gets squared.

class MyInit:    #for initialization
  def __init__(self,items):
    self.items=items
    self.index=0

  def __iter__(self):
    return self

  def __next__(self):  # does the operation
    if self.index>=len(self.items):
      raise StopIteration

    current_itr=self.items[self.index]
    x = current_itr**2
    self.index+=1
    return x

obj1 = MyInit([1,2,3,4,5])
itr = iter(obj1)

print(next(itr))
print(next(itr))
print(next(itr))
print(next(itr))
```

## Strings in Python

Open the file poem.txt and print lenght of each line.

Find sum of each line, mean, median and standard deviation of all lines in the file.

Count the number of words in the file

Print the first 1000 characters from poem.txt.

Print the 3rd to 5th words from the file.

```
import statistics

with open('poem.txt','r') as file:
  lines=file.readlines()


line_lengths = [len(line) for line in lines]

print("--- 1. Length of each line ---")
for i, length in enumerate(line_lengths, start=1):
    print(f"Line {i}: {length} characters")



total_sum = sum(line_lengths)
mean_val = statistics.mean(line_lengths)
median_val = statistics.median(line_lengths)
std_dev = statistics.stdev(line_lengths)

print("\n--- 2. Line Length Statistics ---")
print(f"Sum of all line lengths: {total_sum}")
print(f"Mean line length:        {mean_val:.2f}")
print(f"Median line length:      {median_val}")
print(f"Standard Deviation:      {std_dev:.2f}")

with open("poem.txt", "r") as file:
    full_text = file.read()

all_words = full_text.split()

print("\n--- 3. Word Count ---")
print(f"Total number of words: {len(all_words)}")

print("\n--- 4. First 1000 Characters ---")
print(full_text[:1000])

print("\n--- 5. 3rd to 5th Words ---")

words_3_to_5 = all_words[2:5]
print(words_3_to_5)
```

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

Find the average hwy fuel economy across all cars

Use set to return the unique values for the number of cylinders the cars.

```
# cylinders = set(d['cyl'])
import csv

total_cty = 0
row_count = 0

with open('mpg.csv', mode='r') as csvfile:
    reader = csv.DictReader(csvfile)

    for row in reader:

        total_cty += int(row['cty'])
        row_count += 1

if row_count > 0:
    average_cty = total_cty / row_count
    print(f"Average cty fuel economy: {average_cty:.2f}")
else:
    print("The CSV file is empty.")
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```
import csv
cyl_totals = {}
cyl_counts = {}

with open('mpg.csv', mode='r') as csvfile:
    reader = csv.DictReader(csvfile)

    for row in reader:
        cyl = row['cyl']
        cty_str = row['cty']

        if not cyl or not cty_str:
            continue

        cty_val = float(cty_str)

        if cyl not in cyl_totals:
            cyl_totals[cyl] = 0.0
            cyl_counts[cyl] = 0

        cyl_totals[cyl] += cty_val
        cyl_counts[cyl] += 1

print("Cylinders | Average Cty MPG")
print("-" * 27)

for cyl in sorted(cyl_totals.keys()):
    avg_cty = cyl_totals[cyl] / cyl_counts[cyl]
    print(f"{cyl:<9} | {avg_cty:.2f}")
```

Find average miles per gallon in the highway for each class of vehicles

```
import csv

class_totals = {}
class_counts = {}

with open('mpg.csv', mode='r') as csvfile:
    reader = csv.DictReader(csvfile)

    for row in reader:
        vehicle_class = row['class']
        hwy_str = row['hwy']

        if not vehicle_class or not hwy_str:
            continue

        hwy_val = float(hwy_str)

        if vehicle_class not in class_totals:
            class_totals[vehicle_class] = 0.0
            class_counts[vehicle_class] = 0

        class_totals[vehicle_class] += hwy_val
        class_counts[vehicle_class] += 1

print(f"{'Vehicle Class':<15} | {'Average Hwy MPG':<15}")
print("-" * 33)
for vehicle_class in sorted(class_totals.keys()):
    avg_hwy = class_totals[vehicle_class] / class_counts[vehicle_class]
    print(f"{vehicle_class:<15} | {avg_hwy:.2f}")
```
